```python
def draw_min_rect_rectangle(mask_path):
    image = cv2.imread(mask_path)
    print(image.shape)
    thresh = cv2.Canny(image, 128, 256)
 
    contours, hierarchy = cv2.findContours(thresh, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
    img = np.copy(image)
    for cnt in contours:
        x, y, w, h = cv2.boundingRect(cnt)
        # 绘制矩形
        cv2.rectangle(img,  (x, y+h), (x+w, y), (255,255,255),-1)
        # cv2.rectangle(img,  (x, y+h), (x+w, y), (255),-1)

         # 将最小内接矩形填充为白色
    gray_image = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)  # 如果array已经是灰度数据，则不需要转换
    print(img.shape)
    print(type(img))
    cv2.imwrite('z23213.png',img)
```

```python
import numpy as np
from scipy.spatial.distance import cityblock, chebyshev
from scipy.stats import pearsonr

# 生成两个随机二维数组
array1 = np.array([[1, 1, 1, 1],[1, 1, 1, 1]])
array2 = np.array([[1, 1, 0, 0],[1, 1, 1, 0]])
# 余弦相似度
cosine_similarity = np.dot(array1.flatten(), array2.flatten()) / (np.linalg.norm(array1) * np.linalg.norm(array2))

euclidean_distance = np.linalg.norm(array1 - array2)

# 曼哈顿距离
manhattan_distance = cityblock(array1.flatten(), array2.flatten())

# 切比雪夫距离
chebyshev_distance = chebyshev(array1.flatten(), array2.flatten())

# 杰卡德相似系数
jaccard_similarity = len(np.intersect1d(array1, array2)) / len(np.union1d(array1, array2))

# 皮尔逊相关系数
array1_flattened = array1.flatten()
array2_flattened = array2.flatten()
pearson_correlation, _ = pearsonr(array1_flattened, array2_flattened)

# 输出结果
print("数组1:")
print(array1)
print("数组2:")
print(array2)
print("余弦相似度:", cosine_similarity)
print("欧式距离:", euclidean_distance)
print("曼哈顿距离:", manhattan_distance)
print("切比雪夫距离:", chebyshev_distance)
print("杰卡德相似系数:", jaccard_similarity)
print("皮尔逊相关系数:", pearson_correlation)
print(array1*0.3)
```

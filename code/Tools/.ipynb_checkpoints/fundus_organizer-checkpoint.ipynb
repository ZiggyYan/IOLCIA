```python
import os
import random
import glob
```

```python
domain1_image_adr = '../../Data/Fundus-doFE/Domain1/train/ROIs/image'
domain2_image_adr = '../../Data/Fundus-doFE/Domain2/train/ROIs/image'
domain3_image_adr = '../../Data/Fundus-doFE/Domain3/train/ROIs/image'
domain4_image_adr = '../../Data/Fundus-doFE/Domain4/train/ROIs/image'
domain_image_adrs = [domain1_image_adr,domain2_image_adr,domain3_image_adr,domain4_image_adr]
```

```python
domain1_mask_adr = '../../Data/Fundus-doFE/Domain1/train/ROIs/mask'
domain2_mask_adr = '../../Data/Fundus-doFE/Domain2/train/ROIs/mask'
domain3_mask_adr = '../../Data/Fundus-doFE/Domain3/train/ROIs/mask'
domain4_mask_adr = '../../Data/Fundus-doFE/Domain4/train/ROIs/mask'
domain_mask_adrs = [domain1_mask_adr,domain2_mask_adr,domain3_mask_adr,domain4_mask_adr]
```

```python
domain_nums = []
```

```python
for i in range(4):
    directory = domain_image_adrs[i]
    files = os.listdir(directory)
    # 过滤出文件（排除文件夹）
    domain_nums.append(len(files))
```

```python
domain_nums
```

```python
domain_selected_index = []
```

```python
for i in range(4):
    selected_index = []
    while len(selected_index)<domain_nums[i]*0.4:
        random_number = random.randint(0, domain_nums[i]-1)
        if random_number not in selected_index:
            selected_index.append(random_number)
    domain_selected_index.append(selected_index)
```

```python
import shutil
#@save
def copyfile(filename, target_dir):
    """将文件复制到目标目录"""
    os.makedirs(target_dir, exist_ok=True)
    shutil.copy(filename, target_dir)
```

```python
#Valid didvision
for i in range(4):
    path = domain_image_adrs[i]
    images = glob.glob(domain_image_adrs[i] + "/*")
    masks = glob.glob(domain_mask_adrs[i] + "/*")
    for j in range(len(domain_selected_index[i])):
        old_image_path = images[domain_selected_index[i][j]]
        new_image_path = os.path.join(domain_image_adrs[i].replace('image','valid/image'))
        copyfile(old_image_path,new_image_path)
        old_mask_path = masks[domain_selected_index[i][j]]
        new_mask_path = os.path.join(domain_mask_adrs[i].replace('mask','valid/mask'))
        copyfile(old_mask_path,new_mask_path)
```

```python
#Train didvision
for i in range(4):
    path = domain_image_adrs[i]
    images = glob.glob(domain_image_adrs[i] + "/*")
    masks = glob.glob(domain_mask_adrs[i] + "/*")
    for j in range(len(images)):
        if j not in domain_selected_index[i]:
            old_image_path = images[j]
            new_image_path = os.path.join(domain_image_adrs[i].replace('image','train/image'))
            copyfile(old_image_path,new_image_path)
            old_mask_path = masks[j]
            new_mask_path = os.path.join(domain_mask_adrs[i].replace('mask','train/mask'))
            copyfile(old_mask_path,new_mask_path)
```

```python
import rembg
from PIL import Image
import os
```

```python
def erase_background(file_adr):
    img = Image.open(file_adr)#准备修改的图片路径
    img_bg_remove = rembg.remove(img)# 
    # img_bg_remove = img_bg_remove.convert('RGB')
    #如果需要保存为jpg或者jpeg格式则需要使用此行代码，使用后背景为黑色
    img_bg_remove.save(file_adr)#修改后保存的图片路径
```

```python
BUSBRA_adr = '../../Data/BreastUltrasound/BUSBRA/'
BUSBset_adr = '../../Data/BreastUltrasound/BUSBset/'
BUSI_adr = '../../Data/BreastUltrasound/BUSI/'
USG_adr = '../../Data/BreastUltrasound/USG/'
domain_adr =[BUSBRA_adr,BUSBset_adr,BUSI_adr,USG_adr]
```

```python
for i in range(4):
    adr = domain_adr[i]
    train_adr = os.path.join(adr,'train','mask')
    test_adr = os.path.join(adr,'test','mask')
    files = []
    for file in os.listdir(train_adr):
        file_adr = os.path.join(train_adr,file)
        files.append(file_adr)
    for file in os.listdir(test_adr):
        file_adr = os.path.join(test_adr,file)
        files.append(file_adr)
    for j in range(len(files)):
        erase_background(files[j])
```
